```python


import os
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

warnings.filterwarnings("ignore")

pd.set_option("display.max_columns", 100)
pd.set_option("display.width", 160)
pd.set_option("display.float_format", lambda x: f"{x:.4f}")

sns.set_theme(style="whitegrid")
plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["axes.titlesize"] = 14
plt.rcParams["axes.labelsize"] = 12

RANDOM_STATE = 42


# Tous les résultats sont affichés dans Jupyter et sauvegardés automatiquement :

# results/
# └── eda/
#     ├── tables/
#     ├── figures/
#     └── reports/
RESULTS_DIR = Path("../results/eda")
TABLES_DIR = RESULTS_DIR / "tables"
FIGURES_DIR = RESULTS_DIR / "figures"
REPORTS_DIR = RESULTS_DIR / "reports"

TABLES_DIR.mkdir(parents=True, exist_ok=True)
FIGURES_DIR.mkdir(parents=True, exist_ok=True)
REPORTS_DIR.mkdir(parents=True, exist_ok=True)

print(f"✓ Dossier de résultats : {RESULTS_DIR.resolve()}")


def save_dataframe(df, filename, index=False):
    if df is None:
        print(f"⚠ Tableau non sauvegardé : '{filename}' (DataFrame absent).")
        return

    if not hasattr(df, "empty") or df.empty:
        print(f"⚠ Tableau non sauvegardé : '{filename}' (DataFrame vide).")
        return

    filepath = TABLES_DIR / filename
    df.to_csv(filepath, index=index, encoding="utf-8-sig")
    print(f"✓ Tableau sauvegardé : {filepath}")


def save_figure(filename):
    """
    Sauvegarde la figure matplotlib courante dans results/eda/figures.
    À appeler avant plt.show().
    """
    filepath = FIGURES_DIR / filename
    plt.savefig(filepath, dpi=300, bbox_inches="tight")
    print(f"✓ Figure sauvegardée : {filepath}")


#  2. Configuration et chargement des données


DATA_PATH = "../data/raw/KDDTrain+.txt"

NSL_KDD_COLUMNS = [
    "duration", "protocol_type", "service", "flag",
    "src_bytes", "dst_bytes", "land", "wrong_fragment", "urgent",
    "hot", "num_failed_logins", "logged_in", "num_compromised",
    "root_shell", "su_attempted", "num_root", "num_file_creations",
    "num_shells", "num_access_files", "num_outbound_cmds",
    "is_host_login", "is_guest_login", "count", "srv_count",
    "serror_rate", "srv_serror_rate", "rerror_rate", "srv_rerror_rate",
    "same_srv_rate", "diff_srv_rate", "srv_diff_host_rate",
    "dst_host_count", "dst_host_srv_count", "dst_host_same_srv_rate",
    "dst_host_diff_srv_rate", "dst_host_same_src_port_rate",
    "dst_host_srv_diff_host_rate", "dst_host_serror_rate",
    "dst_host_srv_serror_rate", "dst_host_rerror_rate",
    "dst_host_srv_rerror_rate"
]

TARGET_CANDIDATES = [
    "label", "Label", "labels", "class", "Class",
    "target", "attack_type", "attack"
]

DIFFICULTY_CANDIDATES = [
    "difficulty", "difficulty_level", "difficulty_degree", "level"
]


def load_nsl_kdd(path: str) -> pd.DataFrame:
    """
    Charge le dataset NSL-KDD en détectant automatiquement la présence d'un en-tête.
    """
    if not os.path.exists(path):
        raise FileNotFoundError(f"Le fichier est introuvable : {path}")

    df_raw = pd.read_csv(path, sep=None, engine="python", header=None, nrows=5)
    first_row = df_raw.iloc[0].astype(str).str.strip().str.lower().tolist()

    known_names = set(NSL_KDD_COLUMNS + TARGET_CANDIDATES + DIFFICULTY_CANDIDATES)
    has_header = any(value in known_names for value in first_row)

    if has_header:
        df = pd.read_csv(path, sep=None, engine="python", header=0)
        print("En-tête détecté : les noms de colonnes du fichier ont été utilisés.")
    else:
        expected_columns = NSL_KDD_COLUMNS + ["label", "difficulty"]
        df = pd.read_csv(
            path,
            sep=None,
            engine="python",
            header=None,
            names=expected_columns
        )
        print("Aucun en-tête détecté : noms de colonnes standards NSL-KDD appliqués.")

    print(f"Fichier chargé : {path}")
    print(f"Dimensions initiales : {df.shape[0]} lignes et {df.shape[1]} colonnes.")
    return df


df = load_nsl_kdd("../data/raw/KDDTrain+.txt")


# Aperçu des premières lignes
df.head()

# Dimensions, noms de colonnes et types
print("Dimensions du dataset :", df.shape)
print("\nNoms des colonnes :")
print(list(df.columns))

print("\nTypes des données :")
print(df.dtypes)


# 3. Inspection générale des données


df.info()


df.describe(include="all").T


def build_quality_report(dataframe: pd.DataFrame) -> pd.DataFrame:
    """
    Construit un tableau de qualité des données.
    """
    report = pd.DataFrame({
        "type": dataframe.dtypes.astype(str),
        "valeurs_non_nulles": dataframe.notna().sum(),
        "valeurs_manquantes": dataframe.isna().sum(),
        "pourcentage_manquant": dataframe.isna().mean() * 100,
        "valeurs_uniques": dataframe.nunique(dropna=True)
    })

    report["pourcentage_manquant"] = report["pourcentage_manquant"].round(2)
    return report.sort_values("valeurs_manquantes", ascending=False)


quality_report = build_quality_report(df)

print(f"Nombre total de lignes : {df.shape[0]}")
print(f"Nombre total de colonnes : {df.shape[1]}")
print(f"Nombre total de doublons : {df.duplicated().sum()}")

quality_report


# Sauvegarde des informations générales

dataset_shape = pd.DataFrame([
    {"element": "nombre_de_lignes", "valeur": df.shape[0]},
    {"element": "nombre_de_colonnes", "valeur": df.shape[1]}
])

save_dataframe(dataset_shape, "dataset_shape.csv", index=False)
save_dataframe(quality_report, "data_types.csv", index=True)
save_dataframe(quality_report, "missing_values.csv", index=True)

duplicates_report = pd.DataFrame([
    {
        "indicateur": "nombre_de_doublons",
        "valeur": int(df.duplicated().sum())
    },
    {
        "indicateur": "pourcentage_de_doublons",
        "valeur": round(df.duplicated().mean() * 100, 2)
    }
])

save_dataframe(duplicates_report, "duplicates_report.csv", index=False)


#  4. Statistiques descriptives des variables numériques


numeric_columns = df.select_dtypes(include=[np.number]).columns.tolist()

print(f"Nombre de variables numériques détectées : {len(numeric_columns)}")
print(numeric_columns)


def descriptive_statistics(dataframe: pd.DataFrame, columns: list) -> pd.DataFrame:
    """
    Calcule les statistiques descriptives principales.
    """
    if not columns:
        print("Aucune variable numérique disponible.")
        return pd.DataFrame()

    stats = dataframe[columns].describe().T
    stats["median"] = dataframe[columns].median()
    stats["missing"] = dataframe[columns].isna().sum()

    stats = stats[[
        "mean", "median", "min", "25%", "50%", "75%", "max", "std", "missing"
    ]]

    stats.columns = [
        "moyenne", "médiane", "minimum", "Q1",
        "Q2", "Q3", "maximum", "écart-type", "valeurs_manquantes"
    ]

    return stats.round(4)


numeric_summary = descriptive_statistics(df, numeric_columns)
numeric_summary


save_dataframe(numeric_summary, "descriptive_statistics.csv", index=True)
save_dataframe(numeric_summary, "numeric_summary.csv", index=True)


# 5. Identification automatique des variables

def detect_target_column(dataframe: pd.DataFrame) -> str:
    """
    Détecte automatiquement la colonne cible.
    """
    columns = dataframe.columns

    for candidate in TARGET_CANDIDATES:
        if candidate in columns:
            return candidate

    object_columns = dataframe.select_dtypes(
        include=["object", "category"]
    ).columns.tolist()

    if object_columns:
        return object_columns[-1]

    raise ValueError("Aucune colonne cible n'a pu être détectée automatiquement.")


def get_categorical_columns(dataframe: pd.DataFrame, target_column: str) -> list:
    """
    Retourne les colonnes catégorielles, hors variable cible.
    """
    categorical_columns = dataframe.select_dtypes(
        include=["object", "category", "bool"]
    ).columns.tolist()

    if target_column in categorical_columns:
        categorical_columns.remove(target_column)

    return categorical_columns


TARGET_COLUMN = detect_target_column(df)
categorical_columns = get_categorical_columns(df, TARGET_COLUMN)

print(f"Colonne cible détectée : '{TARGET_COLUMN}'")
print(f"Nombre de classes distinctes : {df[TARGET_COLUMN].nunique(dropna=True)}")

print("\nVariables catégorielles :")
print(categorical_columns)

print("\nVariables numériques :")
print(numeric_columns)


categorical_summary = pd.DataFrame({
    "nombre_categories": df[categorical_columns].nunique(dropna=True),
    "valeur_la_plus_frequente": df[categorical_columns].mode().iloc[0],
    "frequence_valeur_la_plus_frequente": df[categorical_columns].apply(
        lambda series: series.value_counts(dropna=False).iloc[0]
    )
})

save_dataframe(categorical_summary, "categorical_summary.csv", index=True)
categorical_summary

# 6. Analyse des variables numériques


PRIORITY_NUMERIC_COLUMNS = [
    "duration", "src_bytes", "dst_bytes", "count", "srv_count",
    "serror_rate", "srv_serror_rate", "rerror_rate", "srv_rerror_rate",
    "same_srv_rate", "diff_srv_rate"
]

selected_numeric_columns = [
    column for column in PRIORITY_NUMERIC_COLUMNS
    if column in numeric_columns
]

remaining_columns = [
    column for column in numeric_columns
    if column not in selected_numeric_columns
]

MAX_NUMERIC_PLOTS = 12
selected_numeric_columns += remaining_columns[
    :MAX_NUMERIC_PLOTS - len(selected_numeric_columns)
]
selected_numeric_columns = selected_numeric_columns[:MAX_NUMERIC_PLOTS]

print("Variables numériques sélectionnées :")
print(selected_numeric_columns)


def plot_histograms(dataframe: pd.DataFrame, columns: list, bins: int = 40) -> None:
    """
    Affiche et sauvegarde les histogrammes des variables numériques sélectionnées.
    """
    if not columns:
        print("Aucune variable numérique à visualiser.")
        return

    n_cols = 3
    n_rows = int(np.ceil(len(columns) / n_cols))

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, 4 * n_rows))
    axes = np.array(axes).reshape(-1)

    for ax, column in zip(axes, columns):
        sns.histplot(
            data=dataframe,
            x=column,
            bins=bins,
            ax=ax,
            color="#2E86AB"
        )
        ax.set_title(f"Distribution de {column}")
        ax.set_xlabel(column)
        ax.set_ylabel("Fréquence")
        ax.grid(True, alpha=0.3)

    for ax in axes[len(columns):]:
        ax.remove()

    plt.tight_layout()
    save_figure("numeric_distributions.png")
    plt.show()


plot_histograms(df, selected_numeric_columns)

# 7. Analyse des distributions asymétriques


SKEWNESS_COLUMNS = [
    column for column in ["duration", "src_bytes", "dst_bytes"]
    if column in numeric_columns
]

print("Variables sélectionnées pour l'analyse d'asymétrie :")
print(SKEWNESS_COLUMNS)

if SKEWNESS_COLUMNS:
    display(df[SKEWNESS_COLUMNS].skew().to_frame("skewness"))
else:
    print("Aucune variable prioritaire disponible.")


def plot_original_vs_log_distribution(dataframe: pd.DataFrame, column: str) -> None:
    """
    Compare la distribution originale et la distribution transformée par log1p.
    """
    values = dataframe[column].dropna()

    if values.empty:
        print(f"La colonne '{column}' ne contient aucune valeur exploitable.")
        return

    if (values < 0).any():
        print(
            f"Transformation log1p non appliquée à '{column}' "
            "car la variable contient des valeurs négatives."
        )
        return

    transformed_values = np.log1p(values)

    # Distribution originale
    plt.figure(figsize=(10, 5))
    sns.histplot(values, bins=50, color="#2E86AB")
    plt.title(f"Distribution originale — {column}")
    plt.xlabel(column)
    plt.ylabel("Fréquence")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()

    save_figure(f"{column}_distribution.png")
    plt.show()

    # Distribution transformée
    plt.figure(figsize=(10, 5))
    sns.histplot(transformed_values, bins=50, color="#A23B72")
    plt.title(f"Distribution transformée — log1p({column})")
    plt.xlabel(f"log1p({column})")
    plt.ylabel("Fréquence")
    plt.grid(True, alpha=0.3)
    plt.tight_layout()

    save_figure(f"{column}_log_distribution.png")
    plt.show()


for column in SKEWNESS_COLUMNS:
    plot_original_vs_log_distribution(df, column)


#  8. Analyse des variables catégorielles


CATEGORICAL_PRIORITY = [
    column for column in ["protocol_type", "service", "flag"]
    if column in categorical_columns
]

print("Variables catégorielles prioritaires détectées :")
print(CATEGORICAL_PRIORITY)


def categorical_frequency_table(
    dataframe: pd.DataFrame,
    column: str,
    top_n: int | None = None
) -> pd.DataFrame:
    """
    Calcule les fréquences absolues et relatives.
    """
    counts = dataframe[column].value_counts(dropna=False)
    percentages = dataframe[column].value_counts(normalize=True, dropna=False) * 100

    table = pd.DataFrame({
        "fréquence": counts,
        "pourcentage": percentages.round(2)
    })

    if top_n is not None:
        table = table.head(top_n)

    return table


def plot_categorical_frequencies(
    dataframe: pd.DataFrame,
    column: str,
    top_n: int | None = None,
    filename: str | None = None
) -> None:
    """
    Affiche et sauvegarde les catégories les plus fréquentes.
    """
    table = categorical_frequency_table(dataframe, column, top_n)

    if table.empty:
        print(f"La colonne '{column}' ne contient aucune valeur exploitable.")
        return

    plt.figure(figsize=(12, 6))
    ax = sns.barplot(
        x=table.index.astype(str),
        y=table["fréquence"],
        color="#2E86AB"
    )

    title = f"Fréquence des catégories — {column}"
    if top_n is not None:
        title += f" (top {min(top_n, len(table))})"

    ax.set_title(title)
    ax.set_xlabel(column)
    ax.set_ylabel("Nombre d'observations")
    plt.xticks(rotation=45, ha="right")
    plt.grid(True, axis="y", alpha=0.3)
    plt.tight_layout()

    if filename is None:
        filename = f"{column}_frequency.png"

    save_figure(filename)
    plt.show()


for column in CATEGORICAL_PRIORITY:
    top_n = 15 if column == "service" else None

    if column == "service":
        filename = "service_top15_frequency.png"
    else:
        filename = f"{column}_frequency.png"

    print(f"\n### Tableau de fréquence — {column}")
    frequency_table = categorical_frequency_table(df, column, top_n)
    display(frequency_table)

    save_dataframe(
        frequency_table,
        f"{column}_frequency.csv",
        index=True
    )

    plot_categorical_frequencies(df, column, top_n, filename)


#  9. Analyse de la variable cible


def analyze_target_distribution(dataframe: pd.DataFrame, target_column: str) -> pd.DataFrame:
    """
    Calcule les effectifs et pourcentages des classes cibles.
    """
    counts = dataframe[target_column].value_counts(dropna=False)
    percentages = dataframe[target_column].value_counts(normalize=True, dropna=False) * 100

    return pd.DataFrame({
        "nombre_observations": counts,
        "pourcentage": percentages.round(2)
    })


target_summary = analyze_target_distribution(df, TARGET_COLUMN)

print(f"Colonne cible analysée : '{TARGET_COLUMN}'")
print(f"Nombre total de classes : {target_summary.shape[0]}")

save_dataframe(target_summary, "target_distribution.csv", index=True)
target_summary


plt.figure(figsize=(14, 6))
ax = sns.barplot(
    x=target_summary.index.astype(str),
    y=target_summary["nombre_observations"],
    color="#2E86AB"
)

ax.set_title(f"Distribution de la variable cible — {TARGET_COLUMN}")
ax.set_xlabel(TARGET_COLUMN)
ax.set_ylabel("Nombre d'observations")
plt.xticks(rotation=45, ha="right")
plt.grid(True, axis="y", alpha=0.3)
plt.tight_layout()

save_figure("target_distribution_bar.png")
plt.show()


# ============================================================
# TARGET DISTRIBUTION — PIE CHART
# ============================================================

MAX_CLASSES_FOR_PIE = 12

# Trier les classes par fréquence
target_counts = target_summary["nombre_observations"].sort_values(
    ascending=False
)

# Conserver les 11 classes les plus fréquentes
top_classes = target_counts.head(MAX_CLASSES_FOR_PIE - 1)

# Regrouper les autres classes
other_count = target_counts.iloc[MAX_CLASSES_FOR_PIE - 1:].sum()

if other_count > 0:
    pie_data = pd.concat([
        top_classes,
        pd.Series({"Other": other_count})
    ])
else:
    pie_data = top_classes

# Créer le graphique
fig, ax = plt.subplots(figsize=(11, 9))

wedges, texts, autotexts = ax.pie(
    pie_data.values,
    labels=pie_data.index.astype(str),
    autopct="%1.1f%%",
    startangle=90,
    counterclock=False,
    pctdistance=0.75,
    textprops={"fontsize": 10},
    wedgeprops={"edgecolor": "white", "linewidth": 1}
)

ax.set_title(
    f"Target Class Distribution — {TARGET_COLUMN}",
    fontsize=15,
    fontweight="bold"
)

ax.axis("equal")

plt.tight_layout()

# Sauvegarde physique
output_path = FIGURES_DIR / "target_distribution_pie.png"

fig.savefig(
    output_path,
    dpi=300,
    bbox_inches="tight"
)

plt.close(fig)

# Vérification
assert output_path.exists(), "Image non générée !"
assert output_path.stat().st_size > 0, "Image vide !"

print("SUCCESS: Pie chart generated")
print("Path:", output_path.resolve())
print("Size:", output_path.stat().st_size, "bytes")


class_counts = target_summary["nombre_observations"]
imbalance_ratio = class_counts.max() / class_counts.min()

majority_class = class_counts.idxmax()
minority_class = class_counts.idxmin()

print("Analyse du déséquilibre des classes :")
print(f"- Classe majoritaire : '{majority_class}' avec {class_counts.max()} observations.")
print(f"- Classe minoritaire : '{minority_class}' avec {class_counts.min()} observations.")
print(f"- Ratio majoritaire / minoritaire : {imbalance_ratio:.2f}.")
print(
    f"- Pourcentage classe majoritaire : "
    f"{target_summary.loc[majority_class, 'pourcentage']:.2f}%."
)
print(
    f"- Pourcentage classe minoritaire : "
    f"{target_summary.loc[minority_class, 'pourcentage']:.2f}%."
)

if imbalance_ratio > 10:
    print("Conclusion : déséquilibre important des classes.")
elif imbalance_ratio > 3:
    print("Conclusion : déséquilibre modéré des classes.")
else:
    print("Conclusion : classes relativement équilibrées.")


#  10. Matrice de corrélation


CORRELATION_THRESHOLD = 0.8
MAX_CORRELATION_FEATURES = 20

print(f"Seuil de corrélation forte : |r| > {CORRELATION_THRESHOLD}")
print(f"Nombre maximal de variables affichées : {MAX_CORRELATION_FEATURES}")


def select_correlation_columns(
    dataframe: pd.DataFrame,
    columns: list,
    max_features: int
) -> list:
    """
    Sélectionne un sous-ensemble lisible de variables numériques.
    Les variables prioritaires sont conservées en priorité, puis les variables
    à plus grande variance sont ajoutées.
    """
    priority_columns = [
        column for column in [
            "duration", "src_bytes", "dst_bytes", "count", "srv_count",
            "serror_rate", "srv_serror_rate", "rerror_rate", "srv_rerror_rate",
            "same_srv_rate", "diff_srv_rate", "dst_host_count", "dst_host_srv_count"
        ]
        if column in columns
    ]

    remaining_columns = [
        column for column in columns
        if column not in priority_columns
    ]

    if remaining_columns:
        variances = dataframe[remaining_columns].var().sort_values(ascending=False)
        remaining_columns = variances.index.tolist()

    return (priority_columns + remaining_columns)[:max_features]


correlation_columns = select_correlation_columns(
    df, numeric_columns, MAX_CORRELATION_FEATURES
)

print("\nVariables sélectionnées pour la corrélation :")
print(correlation_columns)

correlation_matrix = df[correlation_columns].corr(method="pearson")

plt.figure(figsize=(14, 11))
sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    square=True,
    linewidths=0.5,
    cbar_kws={"shrink": 0.8}
)
plt.title("Matrice de corrélation des variables numériques sélectionnées")
plt.tight_layout()

save_figure("correlation_heatmap.png")
plt.show()


def find_strong_correlations(
    correlation_matrix: pd.DataFrame,
    threshold: float
) -> pd.DataFrame:
    """
    Identifie les paires de variables fortement corrélées.
    """
    strong_pairs = []
    columns = correlation_matrix.columns

    for i in range(len(columns)):
        for j in range(i + 1, len(columns)):
            variable_1 = columns[i]
            variable_2 = columns[j]
            value = correlation_matrix.iloc[i, j]

            if abs(value) > threshold:
                strong_pairs.append({
                    "variable_1": variable_1,
                    "variable_2": variable_2,
                    "corrélation": value,
                    "corrélation_absolue": abs(value)
                })

    result = pd.DataFrame(strong_pairs)

    if result.empty:
        print(
            f"Aucune paire ne dépasse le seuil de corrélation absolue "
            f"de {threshold}."
        )
        return result

    return result.sort_values("corrélation_absolue", ascending=False)


strong_correlations = find_strong_correlations(
    correlation_matrix, CORRELATION_THRESHOLD
)

save_dataframe(strong_correlations, "strong_correlations.csv", index=False)
strong_correlations


#  11. Relations entre variables numériques et la cible


TARGET_NUMERIC_COLUMNS = [
    column for column in [
        "duration", "src_bytes", "dst_bytes", "count", "srv_count",
        "serror_rate", "rerror_rate", "same_srv_rate", "diff_srv_rate"
    ]
    if column in numeric_columns
]

MAX_CLASSES_FOR_BOXPLOT = 10

print("Variables numériques sélectionnées :")
print(TARGET_NUMERIC_COLUMNS)
print(f"\nNombre maximal de classes affichées : {MAX_CLASSES_FOR_BOXPLOT}")


def get_top_target_classes(
    dataframe: pd.DataFrame,
    target_column: str,
    max_classes: int
) -> list:
    """
    Retourne les classes cibles les plus fréquentes.
    """
    return dataframe[target_column].value_counts().head(max_classes).index.tolist()


top_target_classes = get_top_target_classes(
    df, TARGET_COLUMN, MAX_CLASSES_FOR_BOXPLOT
)

df_target_subset = df[df[TARGET_COLUMN].isin(top_target_classes)].copy()

print(f"\nClasses conservées : {top_target_classes}")
print(f"Nombre d'observations conservées : {df_target_subset.shape[0]}")


def plot_boxplots_by_target(
    dataframe: pd.DataFrame,
    numeric_column: str,
    target_column: str,
    classes: list
) -> None:
    """
    Affiche et sauvegarde un boxplot d'une variable numérique selon la cible.
    """
    plt.figure(figsize=(14, 6))

    sns.boxplot(
        data=dataframe[dataframe[target_column].isin(classes)],
        x=target_column,
        y=numeric_column,
        showfliers=False,
        color="#2E86AB"
    )

    plt.title(f"Distribution de {numeric_column} selon {target_column}")
    plt.xlabel(target_column)
    plt.ylabel(numeric_column)
    plt.xticks(rotation=45, ha="right")
    plt.grid(True, axis="y", alpha=0.3)
    plt.tight_layout()

    save_figure(f"boxplot_{numeric_column}_by_class.png")
    plt.show()


for column in TARGET_NUMERIC_COLUMNS:
    plot_boxplots_by_target(
        df_target_subset, column, TARGET_COLUMN, top_target_classes
    )


#  12. Détection simple de patterns


def most_frequent_category_by_target(
    dataframe: pd.DataFrame,
    categorical_column: str,
    target_column: str,
    top_n: int = 5
) -> pd.DataFrame:
    """
    Identifie les catégories les plus fréquentes pour chaque classe cible.
    """
    if categorical_column not in dataframe.columns:
        print(f"La colonne '{categorical_column}' n'est pas disponible.")
        return pd.DataFrame()

    table = (
        dataframe
        .groupby(target_column)[categorical_column]
        .agg(lambda series: series.value_counts().head(top_n).index.tolist())
        .reset_index()
    )

    table.columns = [target_column, f"top_{top_n}_{categorical_column}"]
    return table


if "protocol_type" in df.columns:
    protocol_patterns = most_frequent_category_by_target(
        df, "protocol_type", TARGET_COLUMN, top_n=3
    )
    print("Protocoles les plus fréquents par classe :")
    display(protocol_patterns)
else:
    protocol_patterns = pd.DataFrame()
    print("La colonne 'protocol_type' n'est pas disponible.")


if "service" in df.columns:
    service_patterns = most_frequent_category_by_target(
        df, "service", TARGET_COLUMN, top_n=5
    )
    print("Services les plus fréquents par classe :")
    display(service_patterns)
else:
    service_patterns = pd.DataFrame()
    print("La colonne 'service' n'est pas disponible.")


if TARGET_NUMERIC_COLUMNS:
    numeric_patterns = (
        df.groupby(TARGET_COLUMN)[TARGET_NUMERIC_COLUMNS]
        .mean()
        .round(4)
        .sort_values(by=TARGET_NUMERIC_COLUMNS[0], ascending=False)
    )

    print("Moyennes des principales variables numériques par classe :")
    display(numeric_patterns)

    save_dataframe(numeric_patterns, "class_patterns.csv", index=True)
else:
    numeric_patterns = pd.DataFrame()
    print("Aucune variable numérique prioritaire disponible.")


if not numeric_patterns.empty:
    top_classes_for_patterns = get_top_target_classes(
        df, TARGET_COLUMN, MAX_CLASSES_FOR_BOXPLOT
    )

    numeric_patterns_subset = numeric_patterns.loc[
        numeric_patterns.index.intersection(top_classes_for_patterns)
    ]

    if not numeric_patterns_subset.empty:
        normalized_patterns = numeric_patterns_subset.copy()

        for column in normalized_patterns.columns:
            column_min = normalized_patterns[column].min()
            column_max = normalized_patterns[column].max()

            if column_max != column_min:
                normalized_patterns[column] = (
                    (normalized_patterns[column] - column_min) /
                    (column_max - column_min)
                )
            else:
                normalized_patterns[column] = 0.0

        plt.figure(figsize=(14, 7))
        sns.heatmap(
            normalized_patterns,
            annot=True,
            fmt=".2f",
            cmap="viridis",
            linewidths=0.5,
            cbar_kws={"label": "Valeur moyenne normalisée"}
        )
        plt.title(
            "Moyennes normalisées des variables numériques par classe cible\n"
            "(classes les plus fréquentes)"
        )
        plt.xlabel("Variables numériques")
        plt.ylabel(TARGET_COLUMN)
        plt.tight_layout()

        save_figure("numeric_features_by_class.png")
        plt.show()
    else:
        print("Aucune classe fréquente disponible pour la visualisation.")
else:
    print("Aucune variable numérique prioritaire disponible.")


#  13. Détection des outliers avec la méthode IQR



OUTLIER_COLUMNS = [
    column for column in [
        "duration", "src_bytes", "dst_bytes", "count", "srv_count"
    ]
    if column in numeric_columns
]

print("Variables sélectionnées pour la détection des outliers :")
print(OUTLIER_COLUMNS)


def detect_iqr_outliers(dataframe: pd.DataFrame, column: str) -> pd.DataFrame:
    """
    Détecte les outliers avec la méthode IQR.
    """
    values = dataframe[column].dropna()

    if values.empty:
        return pd.DataFrame([{
            "variable": column,
            "Q1": np.nan,
            "Q3": np.nan,
            "IQR": np.nan,
            "borne_inférieure": np.nan,
            "borne_supérieure": np.nan,
            "nombre_outliers": 0,
            "pourcentage_outliers": 0.0
        }])

    q1 = values.quantile(0.25)
    q3 = values.quantile(0.75)
    iqr = q3 - q1

    lower_bound = q1 - 1.5 * iqr
    upper_bound = q3 + 1.5 * iqr

    outlier_mask = (
        (dataframe[column] < lower_bound) |
        (dataframe[column] > upper_bound)
    )

    return pd.DataFrame([{
        "variable": column,
        "Q1": q1,
        "Q3": q3,
        "IQR": iqr,
        "borne_inférieure": lower_bound,
        "borne_supérieure": upper_bound,
        "nombre_outliers": int(outlier_mask.sum()),
        "pourcentage_outliers": outlier_mask.mean() * 100
    }])


if OUTLIER_COLUMNS:
    outlier_report = pd.concat(
        [detect_iqr_outliers(df, column) for column in OUTLIER_COLUMNS],
        ignore_index=True
    )

    outlier_report["pourcentage_outliers"] = outlier_report[
        "pourcentage_outliers"
    ].round(2)

    save_dataframe(outlier_report, "outliers_report.csv", index=False)
    display(outlier_report)
else:
    outlier_report = pd.DataFrame()
    print("Aucune variable prioritaire disponible.")


if OUTLIER_COLUMNS:
    n_cols = 3
    n_rows = int(np.ceil(len(OUTLIER_COLUMNS) / n_cols))

    fig, axes = plt.subplots(n_rows, n_cols, figsize=(16, 4 * n_rows))
    axes = np.array(axes).reshape(-1)

    for ax, column in zip(axes, OUTLIER_COLUMNS):
        sns.boxplot(data=df, y=column, ax=ax, color="#2E86AB")
        ax.set_title(f"Boxplot — {column}")
        ax.set_xlabel("")
        ax.set_ylabel(column)
        ax.grid(True, axis="y", alpha=0.3)

    for ax in axes[len(OUTLIER_COLUMNS):]:
        ax.remove()

    plt.tight_layout()

    save_figure("outliers_boxplots.png")
    plt.show()
else:
    print("Aucune variable prioritaire disponible pour les boxplots.")


LOG_OUTLIER_COLUMNS = [
    column for column in ["duration", "src_bytes", "dst_bytes"]
    if column in numeric_columns
    and (df[column].dropna() >= 0).all()
]

if LOG_OUTLIER_COLUMNS:
    df_log_outliers = df[LOG_OUTLIER_COLUMNS].copy()

    for column in LOG_OUTLIER_COLUMNS:
        df_log_outliers[column] = np.log1p(df_log_outliers[column])

    plt.figure(figsize=(14, 6))
    sns.boxplot(data=df_log_outliers, color="#A23B72")
    plt.title("Boxplots après transformation log1p")
    plt.xlabel("Variables")
    plt.ylabel("Valeur transformée log1p")
    plt.xticks(rotation=45, ha="right")
    plt.grid(True, axis="y", alpha=0.3)
    plt.tight_layout()

    save_figure("outliers_log_boxplots.png")
    plt.show()
else:
    print(
        "Aucune variable parmi duration, src_bytes et dst_bytes n'est "
        "disponible ou adaptée à log1p."
    )


#  14. Résumé final de l’EDA


missing_total = int(df.isna().sum().sum())
missing_columns_count = int((df.isna().sum() > 0).sum())
duplicates_total = int(df.duplicated().sum())

n_numeric = len(numeric_columns)
n_categorical = len(categorical_columns)
n_classes = df[TARGET_COLUMN].nunique(dropna=True)

top_class = target_summary["nombre_observations"].idxmax()
top_class_percentage = target_summary.loc[top_class, "pourcentage"]

high_correlation_pairs_count = (
    0 if strong_correlations.empty else len(strong_correlations)
)

outlier_variables_count = (
    0 if outlier_report.empty
    else int((outlier_report["nombre_outliers"] > 0).sum())
)

low_variance_columns = [
    column for column in numeric_columns
    if df[column].nunique(dropna=True) <= 1
]

print("Éléments calculés pour le résumé final :")
print(f"- Dimensions : {df.shape[0]} lignes, {df.shape[1]} colonnes.")
print(f"- Valeurs manquantes totales : {missing_total}.")
print(f"- Colonnes avec valeurs manquantes : {missing_columns_count}.")
print(f"- Doublons : {duplicates_total}.")
print(f"- Variables numériques : {n_numeric}.")
print(f"- Variables catégorielles : {n_categorical}.")
print(f"- Colonne cible : '{TARGET_COLUMN}'.")
print(f"- Nombre de classes cibles : {n_classes}.")
print(f"- Classe majoritaire : '{top_class}' ({top_class_percentage:.2f}%).")
print(f"- Paires fortement corrélées : {high_correlation_pairs_count}.")
print(f"- Variables avec outliers IQR : {outlier_variables_count}.")
print(f"- Variables numériques à variance nulle : {low_variance_columns}.")


def build_eda_summary() -> str:
    """
    Construit un résumé textuel basé uniquement sur les résultats calculés.
    """
    lines = []
    lines.append("RÉSUMÉ AUTOMATIQUE DE L'EDA — NSL-KDD / KDDTrain")
    lines.append("=" * 60)

    lines.append("\n1. Dimensions du dataset")
    lines.append(f"- Nombre de lignes : {df.shape[0]}")
    lines.append(f"- Nombre de colonnes : {df.shape[1]}")

    lines.append("\n2. Qualité des données")
    lines.append(f"- Valeurs manquantes totales : {missing_total}")
    lines.append(f"- Colonnes contenant des valeurs manquantes : {missing_columns_count}")
    lines.append(f"- Nombre de doublons : {duplicates_total}")

    lines.append("\n3. Structure des variables")
    lines.append(f"- Variables numériques : {n_numeric}")
    lines.append(f"- Variables catégorielles : {n_categorical}")
    lines.append(f"- Colonne cible détectée : {TARGET_COLUMN}")

    lines.append("\n4. Distribution de la variable cible")
    lines.append(f"- Nombre de classes : {n_classes}")

    target_counts = df[TARGET_COLUMN].value_counts(dropna=False)
    target_percentages = df[TARGET_COLUMN].value_counts(normalize=True, dropna=False) * 100

    for class_name, count in target_counts.items():
        lines.append(
            f"- {class_name} : {count} observations "
            f"({target_percentages[class_name]:.2f}%)"
        )

    lines.append("\n5. Déséquilibre des classes")
    lines.append(f"- Classe majoritaire : {majority_class}")
    lines.append(f"- Classe minoritaire : {minority_class}")
    lines.append(f"- Ratio majoritaire / minoritaire : {imbalance_ratio:.2f}")

    lines.append("\n6. Corrélations fortes")
    if not strong_correlations.empty:
        lines.append(
            f"- Nombre de paires avec |corrélation| > "
            f"{CORRELATION_THRESHOLD} : {len(strong_correlations)}"
        )

        for _, row in strong_correlations.head(10).iterrows():
            lines.append(
                f"- {row['variable_1']} / {row['variable_2']} : "
                f"{row['corrélation']:.4f}"
            )
    else:
        lines.append(
            f"- Aucune paire avec |corrélation| > {CORRELATION_THRESHOLD} "
            "parmi les variables numériques analysées."
        )

    lines.append("\n7. Outliers détectés par la méthode IQR")
    if not outlier_report.empty:
        for _, row in outlier_report.iterrows():
            lines.append(
                f"- {row['variable']} : {row['nombre_outliers']} outliers "
                f"({row['pourcentage_outliers']:.2f}%)"
            )
    else:
        lines.append("- Aucun rapport d'outliers disponible.")

    lines.append("\n8. Patterns principaux par classe")
    if not numeric_patterns.empty:
        lines.append(
            "- Moyennes des principales variables numériques par classe : "
            "voir results/eda/tables/class_patterns.csv"
        )

    if not protocol_patterns.empty:
        lines.append(
            "- Protocoles les plus fréquents par classe : "
            "voir la section 'Détection simple de patterns'."
        )

    if not service_patterns.empty:
        lines.append(
            "- Services les plus fréquents par classe : "
            "voir la section 'Détection simple de patterns'."
        )

    lines.append("\n9. Implications pour le preprocessing")
    lines.append(
        "- Les valeurs manquantes, si présentes, devront être traitées avant l'entraînement."
    )
    lines.append(
        "- Les variables catégorielles devront être encodées lors du preprocessing."
    )
    lines.append(
        "- Les distributions asymétriques pourront nécessiter une transformation adaptée."
    )
    lines.append(
        "- Le déséquilibre des classes devra être pris en compte dans l'évaluation du modèle."
    )
    lines.append(
        "- Les outliers ne doivent pas être supprimés automatiquement, car ils peuvent "
        "correspondre à des comportements réseau anormaux ou à des attaques."
    )

    return "\n".join(lines)


eda_summary_text = build_eda_summary()

summary_path = REPORTS_DIR / "eda_summary.txt"
summary_path.write_text(eda_summary_text, encoding="utf-8")

print(f"✓ Résumé EDA sauvegardé : {summary_path}")
print("\n" + eda_summary_text)
```
