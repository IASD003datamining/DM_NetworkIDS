```python

# ============================================================
# PROJET DM_NETWORKIDS - ANALYSE STATISTIQUE
# Dataset : NSL-KDD - KDDTrain+.txt
# Tests : Chi2, ANOVA, Levene, Kruskal, Correlations
# ============================================================

# ============================================================
# 1. IMPORTATION DES BIBLIOTHEQUES
# ============================================================

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import json
import warnings

from pathlib import Path
from scipy.stats import (
    chi2_contingency,
    f_oneway,
    levene,
    kruskal,
    pearsonr,
    spearmanr
)
from sklearn.preprocessing import LabelEncoder

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (10, 6)
plt.rcParams["figure.dpi"] = 100

print("Bibliotheques importees avec succes !")


# ============================================================
# 2. CONFIGURATION DU PROJET
# ============================================================

PROJECT_ROOT = Path.cwd()

if PROJECT_ROOT.name.lower() == "notebooks":
    PROJECT_ROOT = PROJECT_ROOT.parent

DATASET_CANDIDATES = [
    PROJECT_ROOT / "data" / "raw" / "KDDTrain+.txt",
    PROJECT_ROOT / "data" / "KDDTrain+.csv",
    PROJECT_ROOT / "data" / "raw" / "KDDTrain+.csv",
    PROJECT_ROOT / "data" / "KDDTrain.csv"
]

DATASET_PATH = next(
    (p for p in DATASET_CANDIDATES if p.exists()),
    None
)

if DATASET_PATH is None:
    raise FileNotFoundError(
        "Dataset introuvable. Verifiez le dossier data/raw/"
    )

# Dossiers de resultats
RESULTS_DIR = PROJECT_ROOT / "results" / "statistique"
FIGURES_DIR = RESULTS_DIR / "figures"
TABLES_DIR = RESULTS_DIR / "tables"

for folder in [RESULTS_DIR, FIGURES_DIR, TABLES_DIR]:
    folder.mkdir(parents=True, exist_ok=True)


# ============================================================
# 3. DEFINITION DES COLONNES
# ============================================================

FEATURES = [
    "duration", "protocol_type", "service", "flag",
    "src_bytes", "dst_bytes", "land", "wrong_fragment",
    "urgent", "hot", "num_failed_logins", "logged_in",
    "num_compromised", "root_shell", "su_attempted",
    "num_root", "num_file_creations", "num_shells",
    "num_access_files", "num_outbound_cmds",
    "is_host_login", "is_guest_login", "count",
    "srv_count", "serror_rate", "srv_serror_rate",
    "rerror_rate", "srv_rerror_rate", "same_srv_rate",
    "diff_srv_rate", "srv_diff_host_rate",
    "dst_host_count", "dst_host_srv_count",
    "dst_host_same_srv_rate", "dst_host_diff_srv_rate",
    "dst_host_same_src_port_rate",
    "dst_host_srv_diff_host_rate",
    "dst_host_serror_rate", "dst_host_srv_serror_rate",
    "dst_host_rerror_rate", "dst_host_srv_rerror_rate"
]

COLUMNS_42 = FEATURES + ["label"]
COLUMNS_43 = FEATURES + ["label", "difficulty"]


# ============================================================
# 4. CHARGEMENT DU DATASET
# ============================================================

df = pd.read_csv(
    DATASET_PATH,
    header=None,
    skipinitialspace=True,
    low_memory=False
)

df = df.dropna(axis=1, how="all")

if df.shape[1] == 43:
    df.columns = COLUMNS_43
elif df.shape[1] == 42:
    df.columns = COLUMNS_42
else:
    raise ValueError(
        f"Nombre de colonnes inattendu : {df.shape[1]}"
    )

# Nettoyage des espaces
for col in df.select_dtypes(include=["object", "string"]).columns:
    df[col] = df[col].astype("string").str.strip()

if "label" not in df.columns:
    raise ValueError("Colonne label introuvable.")

print("\nDATASET CHARGE AVEC SUCCES")
print("Chemin :", DATASET_PATH)
print("Dimensions :", df.shape)

display(df.head())


# ============================================================
# 5. PREPARATION DES VARIABLES
# ============================================================

# Cible binaire :
# 0 = Normal
# 1 = Attack

df["is_attack"] = (
    df["label"].str.lower() != "normal"
).astype(int)

# Variables numeriques
numeric_features = [
    col for col in df.select_dtypes(include=np.number).columns
    if col not in ["difficulty", "is_attack"]
]

print("Nombre de variables numeriques :", len(numeric_features))
print("Nombre de labels :", df["label"].nunique())

print("\nDistribution de la cible :")
display(df["is_attack"].value_counts().to_frame("Effectif"))


# ============================================================
# 6. TEST DU CHI-CARRE (CHI2)
# protocol_type vs label
# ============================================================

print("\n" + "=" * 65)
print("6. TEST DU CHI-CARRE : PROTOCOL_TYPE VS LABEL")
print("=" * 65)

# Hypotheses :
# H0 : protocol_type et label sont independants.
# H1 : protocol_type et label sont associes.

alpha = 0.05

chi2_result = {}

data_chi2 = df[
    ["protocol_type", "label"]
].dropna()

contingency = pd.crosstab(
    data_chi2["protocol_type"],
    data_chi2["label"]
)

display(contingency)

if contingency.shape[0] >= 2 and contingency.shape[1] >= 2:

    chi2, p_value, dof, expected = chi2_contingency(
        contingency
    )

    n = contingency.to_numpy().sum()
    k = min(contingency.shape)

    # V de Cramer
    cramers_v = np.sqrt(
        chi2 / (n * (k - 1))
    )

    # Frequences attendues
    expected_df = pd.DataFrame(
        expected,
        index=contingency.index,
        columns=contingency.columns
    )

    min_expected = expected.min()
    proportion_below_5 = (expected < 5).mean()

    decision = (
        "Rejet de H0"
        if p_value < alpha
        else "Non-rejet de H0"
    )

    conclusion = (
        "Association statistiquement significative."
        if p_value < alpha
        else "Pas de preuve suffisante d'une association."
    )

    assumptions_ok = (
        min_expected >= 1 and proportion_below_5 <= 0.20
    )

    chi2_result = {
        "Test": "Chi-square",
        "Variable_1": "protocol_type",
        "Variable_2": "label",
        "Observations": int(n),
        "Chi2_statistic": float(chi2),
        "p_value": float(p_value),
        "Degrees_of_freedom": int(dof),
        "Alpha": alpha,
        "Decision": decision,
        "Conclusion": conclusion,
        "Cramers_V": float(cramers_v),
        "Minimum_expected": float(min_expected),
        "Proportion_expected_below_5": float(proportion_below_5),
        "Assumptions_respected": bool(assumptions_ok)
    }

    print(f"Chi2 statistic : {chi2:.6f}")
    print(f"P-value : {p_value:.10g}")
    print(f"Degrees of freedom : {dof}")
    print(f"V de Cramer : {cramers_v:.6f}")
    print(f"Frequence attendue minimale : {min_expected:.4f}")

    print("\nDecision :", decision)
    print("Conclusion :", conclusion)

    if assumptions_ok:
        print("Conditions pratiques des frequences attendues respectees.")
    else:
        print("Attention : frequences attendues faibles. "
              "Interpreter le Chi2 avec prudence.")

    display(expected_df.round(2))

    contingency.to_csv(
        TABLES_DIR / "chi2_contingency.csv"
    )

    expected_df.to_csv(
        TABLES_DIR / "chi2_expected.csv"
    )

else:
    print("Test Chi2 impossible : categories insuffisantes.")


# ============================================================
# 7. VISUALISATION DU TEST CHI2
# ============================================================

print("\nVISUALISATION DU TEST CHI2")

contingency.plot(
    kind="bar",
    stacked=True,
    figsize=(11, 6)
)

plt.title("Distribution des labels selon protocol_type")
plt.xlabel("Protocol type")
plt.ylabel("Effectif")
plt.legend(title="Label", bbox_to_anchor=(1.02, 1))
plt.tight_layout()

plt.savefig(
    FIGURES_DIR / "chi2_protocol_label.png"
)

plt.show()

# Pourcentages
contingency_percent = pd.crosstab(
    data_chi2["protocol_type"],
    data_chi2["label"],
    normalize="index"
) * 100

contingency_percent.to_csv(
    TABLES_DIR / "chi2_protocol_label_percentages.csv"
)


# ============================================================
# 8. ANOVA : DURATION SELON LABEL
# ============================================================

print("\n" + "=" * 65)
print("8. ANOVA : DURATION SELON LABEL")
print("=" * 65)

# H0 : toutes les moyennes de duration sont egales.
# H1 : au moins une moyenne est differente.

groups = [
    group["duration"].dropna().values
    for _, group in df.groupby("label")
]

groups = [g for g in groups if len(g) > 1]

group_stats = df.groupby("label")["duration"].agg(
    ["count", "mean", "median", "std", "min", "max"]
)

display(group_stats)

group_stats.to_csv(
    TABLES_DIR / "anova_group_statistics.csv"
)

if len(groups) >= 2:

    f_statistic, anova_p = f_oneway(*groups)

    # Taille d'effet eta carre
    grand_mean = df["duration"].mean()

    ss_between = sum(
        len(g) * (np.mean(g) - grand_mean) ** 2
        for g in groups
    )

    ss_total = np.sum(
        (df["duration"].dropna() - grand_mean) ** 2
    )

    eta_squared = (
        ss_between / ss_total
        if ss_total > 0 else np.nan
    )

    anova_decision = (
        "Rejet de H0"
        if anova_p < alpha
        else "Non-rejet de H0"
    )

    anova_conclusion = (
        "Au moins une moyenne de duration differe significativement."
        if anova_p < alpha
        else "Aucune difference significative detectee entre les moyennes."
    )

    anova_result = {
        "Test": "One-way ANOVA",
        "Variable": "duration",
        "Grouping_variable": "label",
        "F_statistic": float(f_statistic),
        "p_value": float(anova_p),
        "Alpha": alpha,
        "Eta_squared": float(eta_squared),
        "Decision": anova_decision,
        "Conclusion": anova_conclusion
    }

    print(f"F-statistic : {f_statistic:.6f}")
    print(f"P-value : {anova_p:.10g}")
    print(f"Eta squared : {eta_squared:.6f}")

    print("\nDecision :", anova_decision)
    print("Conclusion :", anova_conclusion)

else:
    anova_result = {}
    print("ANOVA impossible : groupes insuffisants.")


# ============================================================
# 9. VERIFICATION DE L'HOMOGENEITE DES VARIANCES
# TEST DE LEVENE
# ============================================================

print("\n" + "=" * 65)
print("9. TEST DE LEVENE")
print("=" * 65)

# H0 : les variances des groupes sont egales.
# H1 : au moins une variance est differente.

if len(groups) >= 2:

    levene_stat, levene_p = levene(
        *groups,
        center="median"
    )

    levene_decision = (
        "Variances homogenes"
        if levene_p >= alpha
        else "Variances non homogenes"
    )

    levene_result = {
        "Test": "Levene",
        "Statistic": float(levene_stat),
        "p_value": float(levene_p),
        "Decision": levene_decision
    }

    print(f"Levene statistic : {levene_stat:.6f}")
    print(f"P-value : {levene_p:.10g}")
    print("Decision :", levene_decision)

    if levene_p < alpha:
        print("Attention : l'hypothese d'homogeneite des variances "
              "est rejetee.")
    else:
        print("L'hypothese d'homogeneite des variances n'est pas rejetee.")

else:
    levene_result = {}


# ============================================================
# 10. TEST NON PARAMETRIQUE DE KRUSKAL-WALLIS
# ============================================================

print("\n" + "=" * 65)
print("10. TEST DE KRUSKAL-WALLIS")
print("=" * 65)

# Alternative non parametrique a l'ANOVA.
# H0 : les distributions des groupes sont identiques.
# H1 : au moins une distribution differe.

if len(groups) >= 2:

    kruskal_stat, kruskal_p = kruskal(*groups)

    kruskal_decision = (
        "Rejet de H0"
        if kruskal_p < alpha
        else "Non-rejet de H0"
    )

    kruskal_result = {
        "Test": "Kruskal-Wallis",
        "Statistic": float(kruskal_stat),
        "p_value": float(kruskal_p),
        "Decision": kruskal_decision
    }

    print(f"Kruskal statistic : {kruskal_stat:.6f}")
    print(f"P-value : {kruskal_p:.10g}")
    print("Decision :", kruskal_decision)

else:
    kruskal_result = {}


# ============================================================
# 11. VISUALISATION ANOVA
# ============================================================

print("\nVISUALISATION ANOVA")

plt.figure(figsize=(14, 7))

sns.boxplot(
    data=df,
    x="label",
    y="duration",
    showfliers=False
)

plt.title("Distribution de duration selon les labels")
plt.xlabel("Label")
plt.ylabel("Duration")
plt.xticks(rotation=90)
plt.tight_layout()

plt.savefig(
    FIGURES_DIR / "anova_duration_label.png"
)

plt.show()

# Moyenne de duration par label
duration_means = df.groupby(
    "label"
)["duration"].mean().sort_values(ascending=False)

plt.figure(figsize=(13, 6))

sns.barplot(
    x=duration_means.index,
    y=duration_means.values
)

plt.title("Moyenne de duration selon les labels")
plt.xlabel("Label")
plt.ylabel("Moyenne de duration")
plt.xticks(rotation=90)
plt.tight_layout()

plt.savefig(
    FIGURES_DIR / "duration_means_by_label.png"
)

plt.show()


# ============================================================
# 12. CORRELATIONS AVEC LA CIBLE BINAIRE
# ============================================================

print("\n" + "=" * 65)
print("12. CORRELATIONS AVEC LA CIBLE BINAIRE")
print("=" * 65)

# is_attack : 0 = Normal, 1 = Attack
# Pearson avec une cible binaire correspond a une
# correlation point-biseriale.

correlation_results = []

for col in numeric_features:

    data = df[[col, "is_attack"]].dropna()

    if data[col].nunique() > 1 and data["is_attack"].nunique() > 1:

        pearson_corr, pearson_p = pearsonr(
            data[col], data["is_attack"]
        )

        spearman_corr, spearman_p = spearmanr(
            data[col], data["is_attack"]
        )

        correlation_results.append({
            "Variable": col,
            "Pearson_correlation": pearson_corr,
            "Pearson_p_value": pearson_p,
            "Spearman_correlation": spearman_corr,
            "Spearman_p_value": spearman_p
        })

correlation_df = pd.DataFrame(correlation_results)

correlation_df["Abs_Pearson"] = (
    correlation_df["Pearson_correlation"].abs()
)

correlation_df = correlation_df.sort_values(
    "Abs_Pearson",
    ascending=False
)

display(correlation_df)

correlation_df.to_csv(
    TABLES_DIR / "correlations_target.csv",
    index=False
)

# Graphique Top 15
top_corr = correlation_df.head(15).sort_values(
    "Pearson_correlation"
)

plt.figure(figsize=(11, 7))

sns.barplot(
    data=top_corr,
    x="Pearson_correlation",
    y="Variable"
)

plt.axvline(0, color="black", linewidth=1)
plt.title("Top 15 correlations avec is_attack")
plt.xlabel("Correlation de Pearson")
plt.ylabel("Variable")
plt.tight_layout()

plt.savefig(
    FIGURES_DIR / "correlations_target.png"
)

plt.show()


# ============================================================
# 13. MATRICE DE CORRELATION
# ============================================================

print("\n" + "=" * 65)
print("13. MATRICE DE CORRELATION")
print("=" * 65)

correlation_matrix = df[
    numeric_features + ["is_attack"]
].corr()

plt.figure(figsize=(18, 14))

sns.heatmap(
    correlation_matrix,
    cmap="coolwarm",
    center=0,
    linewidths=0.2
)

plt.title("Matrice de correlation des variables numeriques")
plt.tight_layout()

plt.savefig(
    FIGURES_DIR / "matrice_correlation.png"
)

plt.show()

correlation_matrix.to_csv(
    TABLES_DIR / "matrice_correlation.csv"
)


# ============================================================
# 14. ANALYSE DES VARIABLES CATEGORIELLES
# ============================================================

print("\n" + "=" * 65)
print("14. ASSOCIATION DES VARIABLES CATEGORIELLES")
print("=" * 65)

categorical_results = []

for col in ["protocol_type", "service", "flag"]:

    contingency_table = pd.crosstab(
        df[col],
        df["label"]
    )

    if (
        contingency_table.shape[0] >= 2
        and contingency_table.shape[1] >= 2
    ):

        stat, p, dof, expected = chi2_contingency(
            contingency_table
        )

        n = contingency_table.to_numpy().sum()
        k = min(contingency_table.shape)

        v = np.sqrt(stat / (n * (k - 1)))

        categorical_results.append({
            "Variable": col,
            "Chi2": stat,
            "p_value": p,
            "Degrees_of_freedom": dof,
            "Cramers_V": v
        })

categorical_results_df = pd.DataFrame(
    categorical_results
)

display(categorical_results_df)

categorical_results_df.to_csv(
    TABLES_DIR / "tests_categoriels.csv",
    index=False
)


# ============================================================
# 15. SYNTHESE DES TESTS STATISTIQUES
# ============================================================

print("\n" + "=" * 65)
print("15. SYNTHESE DES TESTS STATISTIQUES")
print("=" * 65)

results = {
    "Chi2": chi2_result,
    "ANOVA": anova_result,
    "Levene": levene_result,
    "Kruskal_Wallis": kruskal_result
}

for test_name, result in results.items():

    print("\nTest :", test_name)

    if result:
        for key, value in result.items():
            print(f"{key} : {value}")
    else:
        print("Test non execute.")

# Sauvegarde JSON
with open(
    RESULTS_DIR / "rapport_statistique.json",
    "w",
    encoding="utf-8"
) as f:

    json.dump(
        results,
        f,
        indent=4,
        ensure_ascii=False
    )

# Tableau recapitulatif des tests
summary_rows = []

if chi2_result:
    summary_rows.append({
        "Test": "Chi2",
        "Statistique": chi2_result["Chi2_statistic"],
        "P_value": chi2_result["p_value"],
        "Decision": chi2_result["Decision"]
    })

if anova_result:
    summary_rows.append({
        "Test": "ANOVA",
        "Statistique": anova_result["F_statistic"],
        "P_value": anova_result["p_value"],
        "Decision": anova_result["Decision"]
    })

if levene_result:
    summary_rows.append({
        "Test": "Levene",
        "Statistique": levene_result["Statistic"],
        "P_value": levene_result["p_value"],
        "Decision": levene_result["Decision"]
    })

if kruskal_result:
    summary_rows.append({
        "Test": "Kruskal-Wallis",
        "Statistique": kruskal_result["Statistic"],
        "P_value": kruskal_result["p_value"],
        "Decision": kruskal_result["Decision"]
    })

summary_df = pd.DataFrame(summary_rows)

display(summary_df)

summary_df.to_csv(
    TABLES_DIR / "synthese_tests_statistiques.csv",
    index=False
)


# ============================================================
# 16. EXPORT FINAL
# ============================================================

print("\n" + "=" * 65)
print("ANALYSE STATISTIQUE TERMINEE AVEC SUCCES !")
print("=" * 65)

print("\nDossier principal :", RESULTS_DIR)
print("Dossier graphiques :", FIGURES_DIR)
print("Dossier tableaux :", TABLES_DIR)

print("\nFichiers generes :")

for path in sorted(RESULTS_DIR.rglob("*")):
    if path.is_file():
        print("-", path.relative_to(PROJECT_ROOT))

print("\nTous les tests statistiques et les visualisations sont termines.")
```
